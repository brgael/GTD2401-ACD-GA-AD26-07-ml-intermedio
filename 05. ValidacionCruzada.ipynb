# Validación Cruzada

En este JNB se revisará el tema de Validación Cruzada.

```python

import pandas as pd

url = 'https://raw.githubusercontent.com/aisheahmedova/melbourne-housing/main/melb_data.csv'

melbourne_data = pd.read_csv(url)

y = melbourne_data['Price']

X = melbourne_data.drop(['Price'], axis=1)

numerical_cols = X.select_dtypes(include=['int64', 'float64']).columns

X = X[numerical_cols].copy()

print("Datos cargados correctamente")
print(X.head())

```

```python

from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

my_pipeline = Pipeline(steps=[
    ('preprocessor', SimpleImputer()),
    ('model', RandomForestRegressor(
        n_estimators=50,
        random_state=0
    ))
])
```

```python

from sklearn.model_selection import cross_val_score

# Multiply by -1 since sklearn calculates *negative* MAE
scores = -1 * cross_val_score(
    my_pipeline,
    X,
    y,
    cv=5,
    scoring='neg_mean_absolute_error'
)

print("MAE scores:\n", scores)
```

```python

print("Average MAE score (across experiments):")
print(scores.mean())
```
