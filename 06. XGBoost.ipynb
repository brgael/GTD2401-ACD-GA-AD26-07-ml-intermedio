# Tema XGBoost

En este JNB se revisará el tema de XGBoost.

```python

import pandas as pd
from sklearn.model_selection import train_test_split

url = 'https://raw.githubusercontent.com/aisheahmedova/melbourne-housing/main/melb_data.csv'

melbourne_data = pd.read_csv(url)

y = melbourne_data['Price']

X = melbourne_data.drop(['Price'], axis=1)

numerical_cols = X.select_dtypes(include=['int64', 'float64']).columns

X = X[numerical_cols].copy()

X_train, X_valid, y_train, y_valid = train_test_split(
    X, y, train_size=0.8, test_size=0.2, random_state=0
)

print("Datos cargados correctamente")
print(X_train.head())
```

```python

!pip install xgboost

from xgboost import XGBRegressor
```

```python

my_model = XGBRegressor()

my_model.fit(X_train, y_train)
```

```python

from sklearn.metrics import mean_absolute_error

predictions = my_model.predict(X_valid)

print("Mean Absolute Error: " + str(mean_absolute_error(predictions, y_valid)))
```

```python

my_model = XGBRegressor(n_estimators=500)

my_model.fit(X_train, y_train)
```

```python

my_model = XGBRegressor(
    n_estimators=500,
    early_stopping_rounds=5
)

my_model.fit(
    X_train,
    y_train,
    eval_set=[(X_valid, y_valid)],
    verbose=False
)
```

```python

my_model = XGBRegressor(
    n_estimators=1000,
    learning_rate=0.05,
    early_stopping_rounds=5
)

my_model.fit(
    X_train,
    y_train,
    eval_set=[(X_valid, y_valid)],
    verbose=False
)
```

```python

my_model = XGBRegressor(
    n_estimators=1000,
    learning_rate=0.05,
    n_jobs=4,
    early_stopping_rounds=5
)

my_model.fit(
    X_train,
    y_train,
    eval_set=[(X_valid, y_valid)],
    verbose=False
)
```
