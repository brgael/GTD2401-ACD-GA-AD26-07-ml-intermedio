# Fuga de Datos

En este JNB se revisará el tema de Fuga de Datos

```python
import pandas as pd
from sklearn.pipeline import make_pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

# Read the data
url = 'https://raw.githubusercontent.com/ghgalal/Kaggle-intermediate-ml/master/data/AER_credit_card_data.csv'

data = pd.read_csv(
    url,
    true_values=['yes'],
    false_values=['no']
)

print("Number of rows in the dataset:", data.shape[0])
print(data.head())
```

```python
# Separar la variable objetivo de las variables predictoras
y = data['card']

X = data.drop(['card'], axis=1)

print("Variables X e y creadas correctamente")
print("X:", X.shape)
print("y:", y.shape)
```

```python
# Crear el modelo
my_pipeline = make_pipeline(
    RandomForestClassifier(n_estimators=100, random_state=0)
)

# Validación cruzada
cv_scores = cross_val_score(
    my_pipeline,
    X,
    y,
    cv=5,
    scoring='accuracy'
)

print("Cross-validation accuracy: %f" % cv_scores.mean())
```
